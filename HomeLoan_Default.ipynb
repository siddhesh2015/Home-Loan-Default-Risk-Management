# 10. Feature Engineering

## Aggregation helper function

```python
def aggregate_numeric(df_in, group_col, prefix, aggs=("mean", "max", "min", "sum")):
    """Aggregate every numeric column per group_col and prefix the new column names."""
    id_cols = [c for c in df_in.columns if c.startswith("SK_ID") and c != group_col]
    num = df_in.drop(columns=id_cols + [group_col]).select_dtypes("number")
    out = num.groupby(df_in[group_col]).agg(list(aggs))
    out.columns = [f"{prefix}_{col}_{agg}".upper() for col, agg in out.columns]
    return out.reset_index()
```

##  Helper function: measure ROC-AUC after each added table

```python
import gc
from lightgbm import LGBMClassifier
from sklearn.metrics import roc_auc_score

def quick_auc(frame):
    """Train the same LightGBM model on the given table and return the validation ROC-AUC."""
    Xf = frame.drop(columns=drop_cols)
    m = LGBMClassifier(n_estimators=500, learning_rate=0.05,
                       random_state=RANDOM_STATE, verbose=-1)
    m.fit(Xf.loc[train_idx], y.loc[train_idx])
    return roc_auc_score(y.loc[val_idx], m.predict_proba(Xf.loc[val_idx])[:, 1])

# Log of ROC-AUC after each step (starts with the baseline)
auc_log = {"application only (baseline)": baseline_auc}
```

## First table: bureau

```python
full = df.copy()

bureau = pd.read_csv(f"{DATA_PATH}/bureau.csv")
bureau_agg = aggregate_numeric(bureau, "SK_ID_CURR", "BUREAU")
full = full.merge(bureau_agg, on="SK_ID_CURR", how="left")
full.index = df.index

del bureau, bureau_agg
gc.collect()

print("Shape after adding bureau:", full.shape)
auc_log["+ bureau"] = quick_auc(full)
print(f"ROC-AUC after bureau: {auc_log['+ bureau']:.4f}")
```

## bureau_balance table

```python
# Step 1: monthly history -> one row per bureau credit
bb = pd.read_csv(
    f"{DATA_PATH}/bureau_balance.csv",
    dtype={"SK_ID_BUREAU": "int32", "MONTHS_BALANCE": "int16", "STATUS": "category"},
)
# STATUS: C = closed, X = unknown, 0 = no delay, 1-5 = increasing days past due
bb["STATUS_BAD"] = bb["STATUS"].isin(["1", "2", "3", "4", "5"]).astype("int8")

bb_agg = bb.groupby("SK_ID_BUREAU").agg(
    MONTHS_COUNT=("MONTHS_BALANCE", "count"),
    MONTHS_MIN=("MONTHS_BALANCE", "min"),
    BAD_SUM=("STATUS_BAD", "sum"),
    BAD_MEAN=("STATUS_BAD", "mean"),
).reset_index()

del bb
gc.collect()
```

```python
# Step 2: link each bureau credit to its client, then aggregate per client
bur_map = pd.read_csv(f"{DATA_PATH}/bureau.csv", usecols=["SK_ID_BUREAU", "SK_ID_CURR"])
bb_client = bur_map.merge(bb_agg, on="SK_ID_BUREAU", how="left")
bb_client_agg = aggregate_numeric(bb_client, "SK_ID_CURR", "BB")

full = full.merge(bb_client_agg, on="SK_ID_CURR", how="left")
full.index = df.index

del bb_agg, bur_map, bb_client, bb_client_agg
gc.collect()

print("Shape after adding bureau_balance:", full.shape)
auc_log["+ bureau_balance"] = quick_auc(full)
print(f"ROC-AUC after bureau_balance: {auc_log['+ bureau_balance']:.4f}")
```

## Table: previous_application

```python
prev = pd.read_csv(f"{DATA_PATH}/previous_application.csv")

# Replace the placeholder 365243 in DAYS_* columns with NaN
day_cols = [c for c in prev.columns if c.startswith("DAYS_")]
prev[day_cols] = prev[day_cols].replace(365243, np.nan)

prev_agg = aggregate_numeric(prev, "SK_ID_CURR", "PREV")
full = full.merge(prev_agg, on="SK_ID_CURR", how="left")
full.index = df.index

# Free memory
del prev, prev_agg
gc.collect()

print("Shape after adding previous_application:", full.shape)
auc_log["+ previous_application"] = quick_auc(full)
print(f"ROC-AUC after previous_application: {auc_log['+ previous_application']:.4f}")
```

## Table: POS_CASH_balance

```python
pos = pd.read_csv(f"{DATA_PATH}/POS_CASH_balance.csv")
pos_agg = aggregate_numeric(pos, "SK_ID_CURR", "POS")
full = full.merge(pos_agg, on="SK_ID_CURR", how="left")
full.index = df.index

# Free memory
del pos, pos_agg
gc.collect()

print("Shape after adding POS_CASH_balance:", full.shape)
auc_log["+ POS_CASH_balance"] = quick_auc(full)
print(f"ROC-AUC after POS_CASH_balance: {auc_log['+ POS_CASH_balance']:.4f}")
```

##  credit_card_balance table

```python
cc = pd.read_csv(f"{DATA_PATH}/credit_card_balance.csv")
cc_agg = aggregate_numeric(cc, "SK_ID_CURR", "CC")
full = full.merge(cc_agg, on="SK_ID_CURR", how="left")
full.index = df.index

# Free memory
del cc, cc_agg
gc.collect()

print("Shape after adding credit_card_balance:", full.shape)
auc_log["+ credit_card_balance"] = quick_auc(full)
print(f"ROC-AUC after credit_card_balance: {auc_log['+ credit_card_balance']:.4f}")
```

## Table: installments_payments

```python
inst = pd.read_csv(f"{DATA_PATH}/installments_payments.csv")

# Behaviour features for each instalment
inst["DAYS_PAST_DUE"] = (inst["DAYS_ENTRY_PAYMENT"] - inst["DAYS_INSTALMENT"]).clip(lower=0)
inst["PAYMENT_RATIO"] = inst["AMT_PAYMENT"] / inst["AMT_INSTALMENT"].replace(0, np.nan)
inst["AMT_UNDERPAID"] = (inst["AMT_INSTALMENT"] - inst["AMT_PAYMENT"]).clip(lower=0)

inst_agg = aggregate_numeric(inst, "SK_ID_CURR", "INST")
full = full.merge(inst_agg, on="SK_ID_CURR", how="left")
full.index = df.index

# Free memory
del inst, inst_agg
gc.collect()

print("Shape after adding installments_payments:", full.shape)
auc_log["+ installments_payments"] = quick_auc(full)
print(f"ROC-AUC after installments_payments: {auc_log['+ installments_payments']:.4f}")
```

## New ratio features

```python
# Financial ratios
full["CREDIT_INCOME_RATIO"] = full["AMT_CREDIT"] / full["AMT_INCOME_TOTAL"]
full["ANNUITY_INCOME_RATIO"] = full["AMT_ANNUITY"] / full["AMT_INCOME_TOTAL"]
full["CREDIT_TERM"] = full["AMT_ANNUITY"] / full["AMT_CREDIT"]
full["CREDIT_GOODS_RATIO"] = full["AMT_CREDIT"] / full["AMT_GOODS_PRICE"]
full["INCOME_PER_PERSON"] = full["AMT_INCOME_TOTAL"] / full["CNT_FAM_MEMBERS"].replace(0, np.nan)

# Share of life spent employed
full["EMPLOYED_AGE_RATIO"] = full["YEARS_EMPLOYED"] / full["AGE_YEARS"]

# Combined external scores (NaN if any score is missing)
full["EXT_SOURCE_MEAN"] = full[["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]].mean(axis=1)
full["EXT_SOURCE_PROD"] = full["EXT_SOURCE_1"] * full["EXT_SOURCE_2"] * full["EXT_SOURCE_3"]

print("Shape after adding ratio features:", full.shape)
auc_log["+ ratio features"] = quick_auc(full)
print(f"ROC-AUC after ratio features: {auc_log['+ ratio features']:.4f}")
```

## ROC-AUC after each step

```python
auc_table = pd.DataFrame({"ROC-AUC": pd.Series(auc_log)})
auc_table["Change vs previous step"] = auc_table["ROC-AUC"].diff()
auc_table["Change vs baseline"] = auc_table["ROC-AUC"] - auc_table["ROC-AUC"].iloc[0]
auc_table.round(4)
```

### Summary of Section 10: Feature Engineering
**Key findings**
- Aggregating the six supporting tables per client and adding ratio features grew the feature set from 121 to 405 features.
- Validation ROC-AUC improved from 0.7598 (application table only) to 0.7874 (+0.0276).
- The largest gains came from previous_application (+0.0116) and bureau (+0.0074). POS_CASH_balance, credit_card_balance, installments_payments and the ratio features each added about +0.002 to +0.003.

**Decisions / next steps**
- Keep the full 405-feature set for modelling; re-check bureau_balance features with feature importance after training.
- Differences smaller than about 0.002 come from a single split and seed, so they should not be over-interpreted.
- Compare several models on this feature set and reduce overfitting (fewer trees, early stopping, regularization) in Section 11.

## Evaluation helper

```python
import time
from sklearn.metrics import (roc_auc_score, precision_score, recall_score,
                             f1_score, precision_recall_curve)

results = []
val_proba = {}

def evaluate(name, model, X_tr, y_tr, X_va, y_va):
    start = time.time()
    model.fit(X_tr, y_tr)
    secs = time.time() - start

    proba = model.predict_proba(X_va)[:, 1]
    pred = (proba >= 0.5).astype(int)

    prec, rec, thr = precision_recall_curve(y_va, proba)
    f1_curve = 2 * prec * rec / (prec + rec + 1e-12)
    best = np.argmax(f1_curve[:-1])

    results.append({
        "Model": name,
        "ROC-AUC": roc_auc_score(y_va, proba),
        "Precision@0.5": precision_score(y_va, pred, zero_division=0),
        "Recall@0.5": recall_score(y_va, pred),
        "F1@0.5": f1_score(y_va, pred),
        "Best F1": f1_curve[best],
        "Best threshold": thr[best],
        "Train time (s)": secs,
    })
    val_proba[name] = proba
    return model
```

# 17. New Prediction

## Prediction function

```python
# Probability edges of the ten risk bands, taken from the validation predictions
risk_edges = np.quantile(proba, np.linspace(0, 1, 11))

def make_applicant(base_row, **changes):
    """Copy a template applicant, apply the changes and recalculate the ratio features."""
    a = base_row.copy()
    for col, value in changes.items():
        a[col] = value
    a["CREDIT_INCOME_RATIO"] = a["AMT_CREDIT"] / a["AMT_INCOME_TOTAL"]
    a["ANNUITY_INCOME_RATIO"] = a["AMT_ANNUITY"] / a["AMT_INCOME_TOTAL"]
    a["CREDIT_TERM"] = a["AMT_ANNUITY"] / a["AMT_CREDIT"]
    a["CREDIT_GOODS_RATIO"] = a["AMT_CREDIT"] / a["AMT_GOODS_PRICE"]
    ext = ["EXT_SOURCE_1", "EXT_SOURCE_2", "EXT_SOURCE_3"]
    a["EXT_SOURCE_MEAN"] = a[ext].mean(axis=1)
    a["EXT_SOURCE_PROD"] = a["EXT_SOURCE_1"] * a["EXT_SOURCE_2"] * a["EXT_SOURCE_3"]
    return a

def predict_default(applicant):
    p = float(lgbm.predict_proba(applicant)[:, 1][0])
    band = int(np.searchsorted(risk_edges[1:-1], p, side="right")) + 1
    decision = "High risk: flag for review" if p >= best_thr else "Acceptable risk"
    return {"Default probability": round(p, 3), "Risk band": f"D{band}", "Decision": decision}
```

## Three example applicants

```python
template = X_val.iloc[[0]]
low_risk = make_applicant(template, AMT_INCOME_TOTAL=300000, AMT_CREDIT=400000,
                          AMT_ANNUITY=20000, AMT_GOODS_PRICE=400000,
                          EXT_SOURCE_1=0.75, EXT_SOURCE_2=0.80, EXT_SOURCE_3=0.78)
high_risk = make_applicant(template, AMT_INCOME_TOTAL=90000, AMT_CREDIT=900000,
                           AMT_ANNUITY=45000, AMT_GOODS_PRICE=700000,
                           EXT_SOURCE_1=0.20, EXT_SOURCE_2=0.25, EXT_SOURCE_3=0.15)
pd.DataFrame(
    [predict_default(template), predict_default(low_risk), predict_default(high_risk)],
    index=["Template applicant (unchanged)", "Low-risk example", "High-risk example"],
)
```

### Summary of Section 17: New Prediction
- A prediction function returns the default probability, the risk band (D1 to D10) and a decision at the tuned threshold of 0.172.
- For a low-risk example (high external scores, credit well below income) the model predicts a 1.0% default probability (band D1). For a high-risk example (low external scores, credit ten times income) it predicts 24.2% (band D10) and flags the applicant for review.
- This is a demonstration: the credit-history features come from a template applicant. A real applicant would need data from all seven tables.

# 18. Report on Challenges Faced

## Challenges

1. Strong class imbalance: only 8.07% of applicants default.
2. Many missing values: 67 columns had missing data at the start, up to 70% missing.
3. Missing values that carry meaning: OWN_CAR_AGE is missing exactly for applicants without a car.
4. Fake placeholder values: 365243 in DAYS_EMPLOYED (18% of rows) and "XNA" in text columns.
5. Seven tables with many rows per client; bureau_balance has no client ID of its own.
6. Large data on a 16 GB laptop (about 27 million and 13 million rows in the biggest tables).
7. Overfitting: baseline ROC-AUC of 0.86 on training rows but 0.76 on validation rows.
8. Accuracy was misleading and the 0.5 threshold caught almost no defaulters (recall 0.05).
9. The first model comparison scored about 0.76 instead of the expected 0.79.
10. A sensitive attribute (CODE_GENDER) is among the top features.


## Techniques used

1. Stratified train/validation split; ROC-AUC, precision, recall and F1 instead of accuracy; class weights; threshold tuning.
2. Kept NaN for tree-based models; median imputation and one-hot encoding inside a Pipeline for linear models, fitted on training rows only.
3. Not filled with the mean; kept as NaN for trees and set to 0 for linear models, together with FLAG_OWN_CAR.
4. Replaced the placeholders with NaN, added the flag DAYS_EMPLOYED_ANOM, and applied the same cleaning to previous_application.
5. Aggregated each table per client (mean, max, min, sum) and joined it on SK_ID_CURR; aggregated bureau_balance in two steps (per credit, then per client).
6. Loaded one table at a time with compact data types, deleted objects and freed memory after use, and stored features as float32.
7. Feature subsampling, row subsampling and a larger minimum leaf size in LightGBM.
8. Tuned the decision threshold to 0.172 (best F1) and compared it with class weighting.
9. Checked the matrix shapes, found that the old 121 features had been used instead of the 405 engineered features, rebuilt the matrices and retrained all models.
10. Flagged for a fairness review before production.

## Reasons

1. A model that predicts "no default" for everyone is 91.9% accurate but useless; stratification keeps the same default rate in train and validation.
2. Tree models handle NaN natively, and fitting imputers on training rows only avoids data leakage.
3. A mean would invent a car age that does not exist.
4. The fake values would distort the model, while the flag keeps useful information (this group defaults less: 5.40% vs 8.66%).
5. The model needs exactly one row per client.
6. It prevents memory crashes and keeps the notebook runnable on a normal laptop.
7. These settings stop the model from memorizing the training data; the train-validation gap fell from about 0.099 to 0.077 and the validation score rose.
8. Recall rose to 0.42, and weighting and threshold tuning gave the same best F1, so the threshold matters more than the weighting.
9. New scores should always be compared with earlier ones, and shapes checked after the feature set changes.
10. Using gender in credit decisions is restricted in many regions.

### Summary of Section 18: Challenges
The main challenges were the class imbalance, placeholder and structural missing values, combining seven tables on a memory-limited laptop, and overfitting. Each was handled with a specific technique, and the reasoning is listed above.

# 19. Conclusion and Recommendations

### Conclusion
- Goal: identify the applicants and factors linked to home loan default (TARGET = 1) and build a model that ranks applicants by risk.
- Data: 307,511 applicants with 8.07% defaulters, combined with six supporting tables into 405 features per client.
- Best model for production: LightGBM (validation ROC-AUC 0.7892). XGBoost scored almost the same (0.7895) but trained about 3.5 times slower (187 s vs 53 s).
- Feature engineering raised ROC-AUC from 0.7598 (baseline) to 0.7892, and the train-validation gap fell from about 0.099 to 0.077.
- At the tuned threshold of 0.172 the model catches 42% of defaulters (recall 0.42) with precision 0.29 and F1 0.34.
- The risk bands work well: the default rate rises from 0.94% (D1) to 30.74% (D10), and the riskiest 20% of applicants (D9 and D10) hold about 57% of all defaulters.

###  Key factors linked to default
- External credit scores (EXT_SOURCE_MEAN, EXT_SOURCE_3, EXT_SOURCE_2) are the strongest factors; low scores raise the risk.
- Loan structure: CREDIT_TERM, CREDIT_GOODS_RATIO and AMT_ANNUITY; loans that are large relative to income or goods price are riskier.
- Repayment behaviour: more days past due on instalments (INST_DAYS_PAST_DUE_MEAN) raises the risk.
- Applicant profile: longer employment, higher age and higher education lower the risk; ORGANIZATION_TYPE and OCCUPATION_TYPE also contribute.

###  Recommendations
1. Use the LightGBM score to place applicants in the 10 risk bands instead of a single yes/no answer.
2. Approve D1 to D3 applicants automatically (default rate below 2.2%).
3. Send the middle bands (D4 to D8) for manual review.
4. Reject D9 and D10 applicants or offer stricter terms (higher down payment, lower amount, higher interest).
5. Choose the threshold by business cost: lower it to catch more defaulters, raise it to reject fewer good customers.
6. Review CODE_GENDER for fairness before production, because gender-based credit decisions are restricted in many regions.
7. Monitor the score distribution and default rate per band every month, and retrain when performance drops.

### Limitations
- The threshold (0.172) was chosen on the validation set, so the reported precision and recall may be slightly optimistic; confirm on a separate test set.
- The model learns from one historical period; customer behaviour and the economy change over time.
- External scores are the strongest inputs, so the model depends on them being available.
- Probabilities are good for ranking but are not calibrated for exact default pricing.

###  Future work
- Tune hyperparameters and calibrate the probabilities.
- Train a version without CODE_GENDER and compare the loss in ROC-AUC.
- Add a separate hold-out test set and choose the threshold by cost.
